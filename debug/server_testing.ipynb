```python
from langchain_openai import ChatOpenAI

# Replace "YOUR_SERVER_IP" with the actual IP address of your llama.cpp server
# For example: "192.168.1.100" or "10.0.0.5" etc.
SERVER_IP = "10.88.222.26"  # <-- Change this to your server's IP address
PORT = 8080  # Default llama.cpp port, change if different

llm = ChatOpenAI(
    openai_api_base=f"http://{SERVER_IP}:{PORT}/v1",
    openai_api_key="unused",  # llama.cpp doesn't require an API key
)

resp = llm.invoke("What is the capital of Oman?")
print(resp.content)
```

```python
# Test connection to your llama.cpp server
import requests

# Use the same SERVER_IP and PORT from above
SERVER_IP = "YOUR_SERVER_IP"  # <-- Change this to match your server's IP
PORT = 8080

try:
    # Test if the server is reachable
    response = requests.get(f"http://{SERVER_IP}:{PORT}/v1/models", timeout=10)
    print("✅ Server is reachable!")
    print(f"Status code: {response.status_code}")
    
    if response.status_code == 200:
        models = response.json()
        print("Available models:", models)
    else:
        print("Server responded but with an error")
        print("Response:", response.text)
        
except requests.exceptions.ConnectionError:
    print(f"❌ Cannot connect to server at {SERVER_IP}:{PORT}")
    print("Check that:")
    print("1. The server IP address is correct")
    print("2. The llama.cpp server is running")
    print("3. The port is correct (default is 8080)")
    print("4. There are no firewall issues blocking the connection")
except requests.exceptions.Timeout:
    print(f"❌ Connection timed out to {SERVER_IP}:{PORT}")
    print("The server might be slow to respond or unreachable")
except Exception as e:
    print(f"❌ Error: {e}")
```
